```python
from pathlib import Path
import json
import sys

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data" / "ontologyqa_test_questions_v1.csv").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

sys.path.insert(0, str(PROJECT_ROOT))


from experiments.runners.sparql_self_correct import SparqlSelfCorrectConfig, SparqlSelfCorrectRunner, summarize

PROJECT_ROOT
```

```python
config = SparqlSelfCorrectConfig.from_project_root(PROJECT_ROOT)

print("Dataset:", config.csv_path)
print("Env:", config.env_path)
print("Model:", config.model)
print("Provider only:", config.provider_only)
print("GraphDB:", config.graphdb_endpoint)
print("Max SPARQL attempts:", config.max_sparql_attempts)
print("Output root:", config.output_root)
```

```python
LIMIT = 0
OFFSET = 0

runner = SparqlSelfCorrectRunner(config)
results, output_dir = runner.run(limit=LIMIT, offset=OFFSET)

summary = summarize(results)
print(json.dumps(summary, ensure_ascii=False, indent=2))
print("Saved to:", output_dir)
```
