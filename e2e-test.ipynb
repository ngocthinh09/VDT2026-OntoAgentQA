```python
from pprint import pprint
from agents.ontology_qa_agent import build_ontology_qa_agent

agent = build_ontology_qa_agent(max_iterations=8)

for update in agent.stream(
    {"question": "What is the length of Type 052D destroyer?"},
    stream_mode="updates",
):
    print("\n=== UPDATE ===")
    for node, data in update.items():
        print("NODE:", node)
        if node == "agent":
            msg = data["messages"][-1]
            print("AI:", msg.content)
            if getattr(msg, "tool_calls", None):
                print("TOOL CALLS:")
        else:
            pprint(data)
```
